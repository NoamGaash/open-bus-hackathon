```python
from openbus_hack import stride, use_openbus_style

use_openbus_style()  # shared chart styling, so everyone's charts match

# Who are the operators?
stride.agencies().head(10)
```

```python
# Routes for a line. Note one "line" = many routes (per direction/alternative/day).
routes = stride.routes(lines=["480"])
print(len(routes), "route rows")
routes[["date", "line_ref", "route_short_name", "route_long_name", "agency_name"]].head()
```

```python
# Pre-aggregated planned vs actual — much kinder to the API than paging raw rides.
# Valid group_by fields: gtfs_route_date, gtfs_route_hour, operator_ref, day_of_week, line_ref
d_from, d_to = stride.default_window(7)
agg = stride.gtfs_rides_agg(d_from, d_to, group_by="operator_ref,gtfs_route_date")
agg.head()
```
